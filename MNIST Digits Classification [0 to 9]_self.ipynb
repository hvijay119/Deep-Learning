# Let's build our 1st Artificial Deep Neural Network(ANN)

## 1. Import Necessary Libraries

```python
import keras
from matplotlib import pyplot as plt
import pandas as pd
```

## 2.Data Collection

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

## 3. Data Understanding

```python
X_train.shape, y_train.shape
```

```python
X_test.shape,y_test.shape
```

```python
assert X_train.shape == (60000, 28, 28)
assert X_test.shape == (10000, 28, 28)
assert y_train.shape == (60000,)
assert y_test.shape == (10000,)
```

### 3.1 Let's see how one image looks like

```python
pd.DataFrame(X_train[10000])
```

```python
plt.imshow(X_train[10000],cmap = "grey")
```

## 4. Data preparation

### Let's Normalize this for Faster processing

```python
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
X_train.min(),X_train.max()
```

```python
X_test.min(),X_test.max()
```

## 5. Model building

```python
from keras.models import Sequential
from keras.layers import Flatten,Dense
```

```python
model=Sequential()
model.add(layer=Flatten())
model.add(Dense(units=100,activation='relu'))
model.add(Dense(units=30,activation='relu'))
model.add(Dense(units=10,activation='softmax'))
```

```python
model.compile(optimizer='rmsprop',loss='SparseCategoricalCrossentropy',metrics=['sparse_categorical_accuracy'])
```

## 6. Model Training || 7. Model Testing || 8. Model Evaluation

```python
model.fit(x=X_train, y=y_train,batch_size=32,epochs=20,validation_data = (X_test,y_test))
```

```python
model.summary()
```

## 9. Manual model prediction

```python
y_pred=model.predict(X_test)
```
