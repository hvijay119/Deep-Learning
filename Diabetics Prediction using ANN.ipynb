# Problem Statement
#### Build a machine learning model to accurately predict whether or not the patients in the dataset have diabetes or not?

### ======================================================================

### 1. Import Necessary libraries

```python
import pandas as pd
import keras
from sklearn.preprocessing import StandardScaler
```

### 2. Import Data

```python
diabetes_data = pd.read_csv('diabetes.csv')
diabetes_data
```

### 3. Data Understanding

```python
diabetes_data.shape
```

```python
diabetes_data.isna().sum()
```

```python
diabetes_data.dtypes
```

### 4. Data Preprocessing

```python
X = diabetes_data.drop('Outcome',axis = 1)
y = diabetes_data[['Outcome']]
```

### 5. Model Building

```python
from keras.models import Sequential
from keras.layers import Dense,Dropout

import warnings
warnings.filterwarnings("ignore")
```

```python
model = Sequential()
model.add(Dense(12, input_dim=8, kernel_initializer='uniform', activation='relu'))
model.add(Dense(8,kernel_initializer='uniform', activation='relu'))
model.add(Dense(1, kernel_initializer='uniform', activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=12,stratify=y)
```

```python
X_train.shape,y_train.shape,X_test.shape,y_test.shape
```

### ALERT: Here's where You Need to apply STANDARDIZATION

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)  
X_test = scaler.transform(X_test)   
```

### ======================================================================

### 6. Model Training || 7. Model Testing || 8. Model Evaluation

```python
model.fit(X_train, y_train,epochs=50,batch_size=16,validation_split=0.2)
```

### 9. Model Prediction

```python
y_pred = model.predict(X_test)
y_pred = (y_pred > 0.60).astype(int)
y_pred
```

```python
model.save("diabetes_intelligence.h5")
```

### ======================================================================

## BONUS: Here you can check other Metrics for structured Data...

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

### ========================================================

## 10. NEW INPUT PREDICTION

### Write down the prediction pipeline you did while Training

##### input(array) -> standardization -> model -> result(will be in array) -> so use typecasting -> return the single output

```python
from keras.models import load_model
intelligence_diabetes = load_model("diabetes_intelligence.h5")
```

```python
import numpy as np
new_data = np.array([[6,148,72,35,0,33.6,0.627,50]])
new_data = scaler.transform(new_data)
prob = intelligence_diabetes.predict(new_data)
print("Probability:", prob)
final_class = (prob > 0.5).astype(int)
print("Final Prediction:", final_class[0][0])
```
